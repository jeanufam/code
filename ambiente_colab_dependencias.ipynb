```python
def safe_import(module_name):
    try:
        module = __import__(module_name)
        print(f"OK - {module_name}")
        return module
    except ImportError as e:
        print(f"ERRO - {module_name}")
        print(e)
        return None

argparse = safe_import("argparse")
numpy = safe_import("numpy")
pandas = safe_import("pandas")
xlsxwriter = safe_import("xlsxwriter")
openpyxl = safe_import("openpyxl")
colorama = safe_import("colorama")
tqdm = safe_import("tqdm")
crossref = safe_import("crossref")
```

```python
from crossref.restful import Works

def search_crossref(term, rows=5):
    works = Works()
    results = []
    for item in works.query(bibliographic=term):
        results.append(item)
        if len(results) >= rows:
            break
    return results

for item in search_crossref("molecular mechanism", 2):
    print(item.get("title", ["N/A"])[0])
```
